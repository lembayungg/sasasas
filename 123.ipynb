### Практика 7. ⭐ Главная итоговая практика — Мини-анализ каталога

```python
names = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники", "Коврик для мыши", "Флеш-карта", "Миктофон"]
```

```python
len(names)
```

```python
prices = [99000, 3499, 5799, 18999, 3999, 299, 899, 4299]
```

```python
total_price = sum(prices)
total_price
```

```python
print(min(prices))
```

```python
print(max(prices))
```

```python
print(total_price // len(names))
```

```python
product = {}
products = []
for i in range (0, len(names)):
    product["name"] = names[i]
    product['price'] = prices[i]
    products.append(product)
products
```
