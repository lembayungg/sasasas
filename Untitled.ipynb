```python
products = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники", "Веб-камера"]
```

Первый продукт:

```python
products[0]
```

Последнйи товар:

```python
products[-1]
```

срезом выведите первые три товара:

```python
products[:3]
```

срезом выведите товары со второго по четвёртый:

```python
products[1:4]
```

получите список товаров в обратном порядке:

```python
products[::-1]
```

добавьте "Микрофон" в конец:

```python
products.append("Микрофон")
products
```

вставьте "Коврик" на позицию с индексом 2:

```python
products.insert(2, "Коврик")
products
```

удалите "Мышь":

```python
products.remove('Мышь')
products
```

выведите итоговый список и его длину:

```python
f'Итоговый лист - {products}, его длинна - {len(products)}'
```

практика показала, как обращаться к э-там листа по индексам: как переворачивать лист, как удалять или добавлять э-ты
