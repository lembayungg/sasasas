### Практика 5. ⭐ Обязательная на занятии — Печать каталога

```python
products = ["Ноутбук", "Мышь", "Клавиатура", "Монитор", "Наушники"]
prices = [79990, 1490, 5990, 24990, 7990]
total_price = 0
```

```python
for i in products:
    print(i)
```

```python
for number, product_name in enumerate(products, start=1):
    print(f"{number}. {product_name}")
```

```python
for i in range(1, len(products) + 1):
    print(i)
```

```python
for product_name, price in zip(products, prices):
    print(f"{product_name}: {price} руб.")
```

```python
for i in prices:
    total_price += i
total_price
```

range() нужен, когда нужно пройтись циклом for  определенное количество итераций
enumerate() нужен, когда пронумеровать список
zip() нужен, когда объединить два списка
